```python
import sys


sys.path.append("..")
from transtrans import Annotation

annotation = Annotation.from_json(Path("../data/BV1iddQYQE7D/annotation.json"))
```

```python
tags, counts = np.unique(annotation.tags, return_counts=True)
```

```python
l = list(zip(tags, counts))
```

```python
l = sorted(l, key=lambda x: x[1], reverse=True)
```

```python
def tag_stats(tag_tokens, slice=10):
    tokens, freqs = np.unique(tag_tokens, return_counts=True)
    l = list(zip(tokens, freqs))
    l.sort(key=lambda x: x[1], reverse=True)
    for token, freq in l[:slice]:
        print(f"{token}: {freq}")
```

```python
for tag, count in l:
    tokens = []
    for token_, tag_ in zip(annotation.tokens, annotation.tags):
        if tag_ == tag:
            tokens.append(token_)
    print(f"{tag}: {count}")
    tag_stats(tokens)
    print("----------")
```
